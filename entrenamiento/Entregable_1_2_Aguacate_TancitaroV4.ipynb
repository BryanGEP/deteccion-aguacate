## 2. DESCRIPCION DEL PROBLEMA

### 2.1 Planteamiento del Problema

En el municipio de Tancítaro Michoacán, la industria aguacatera representa el principal motor económico de la región. A pesar de ser la capital mundial del aguacate, los productores enfrentan pérdidas económicas significativas ocasionadas por enfermedades fungosas que afectan la calidad del fruto durante las etapas de producción y postcosecha.

El método tradicional de inspección consiste en la revision visual manual realizada por trabajadores capacitados. Este procedimiento presenta múltiples limitaciones: es lento, costoso, propenso al error humano, poco escalable para grandes volúmenes de producción, y dependiente del inspector. Las enfermedades en etapas tempranas como la antracnosis inicial son frecuentemente confundidas con frutos sanos, lo que permite que frutos enfermos lleguen al mercado o contaminen el resto de la produccion.

- ¿Es posible desarrollar un modelo de Machine Learning capaz de clasificar automáticamente el estado de un fruto de aguacate a partir de una imagen digital con precisión?
- ¿Puede una CNN (Red Neuronal Convolucional) distinguir de manera confiable entre frutos sanos, con antracnosis y con roña?
- ¿Que tan eficaz es el modelo para detectar la antracnosis en etapas tempranas, cuando los síntomas son visualmente similares a los de un fruto sano?

### 2.2 Hipotesis

Un modelo CNN entrenado con imágenes de aguacates etiquetadas puede clasificar el estado del fruto con precisión superior al 80%, lo que lo hace viable como herramienta de control de calidad en la industria aguacatera de Tancítaro.

La antracnosis representa el mayor reto de clasificación debido a la similitud visual de sus síntomas tempranos con frutos sanos, por lo que se esperan métricas más bajas para esta clase.

Un sistema automatizado de detección puede reducir las pérdidas postcosecha en al menos un 30% al eliminar el error humano en la inspección visual.

## 3. MÉTRICAS PARA MEDIR EL EXITO DEL PROYECTO

| Métrica | Descripción | Valor de Aceptación |
|---------|-------------|---------------------|
| **Precisión Global (PG / Accuracy)** | Proporción de predicciones correctas sobre el total. Mide el desempeño general del clasificador sobre las tres clases. | >= 80% |
| **Precisión por clase (P)** | Indica que tan confiable es el modelo cuando predice una clase. | >= 75% para S y R; >= 60% para A |
| **Exhaustividad / Recall (E)** | Mide que tan bien el modelo encuentra todos los casos reales de una clase. | >= 70% para las tres clases |
| **F1-Score** | Media armónica entre Precisión y Recall. Estimador conservador que penaliza desequilibrios entre ambas métricas. Fundamental con desbalance de clases. | >= 0.70 para S y R; >= 0.60 para A |
| **Pérdida / Loss** | Valor de la función de costo. Un valor bajo y convergente entre entrenamiento y validación indica buen aprendizaje sin sobreajuste. | <= 0.35 en validación |

**Nota:** Se establecieron criterio mas flexible para la clase A (Antracnosis) porque sus síntomas tempranos se confunden visualmente con frutos sanos, representando un desafío técnico inherente al problema.

## 5. PROBLEMAS ENCONTRADOS Y SOLUCIONES

### Problema 1: Desbalance entre clases

**Descripción:** El dataset presenta un número desigual de imágenes por clase: Healthy (1,764 = 44.2%), Scab (1,197 = 30.0%) y Anthracnose (1,022 = 25.7%). El ratio de desbalance entre la clase mayoritaria y la minoritaria es de 1.73:1. Cuando existe desbalance en datos de entrenamiento, los modelos tienden a sobreclasificar la clase mayoritaria, lo que produce una precisión global artificialmente alta pero un desempeño deficiente para las clases minoritarias.

**Solución aplicada:**
- **Pesos de clase ponderados (`class_weight`):** Se asignan pesos inversamente proporcionales a la frecuencia de cada clase durante el entrenamiento CNN, penalizando más los errores en la clase Anthracnose.
- **Data Augmentation durante el entrenamiento:** Se generan variaciones sintéticas de las imágenes (rotaciones, volteos, zoom) de forma balanceada para aumentar la diversidad sin requerir imágenes adicionales.

**Justificación:** Según Johnson y Khoshgoftaar (2019), el F1-Score es el estimador más adecuado para clasificadores con desbalance de clases. Los pesos de clase son una solución estándar en TensorFlow/Keras que no requiere modificar el dataset original.

---

### Problema 2: Similitud visual entre antracnosis temprana y fruto sano

**Descripción:** Los frutos con antracnosis en etapa inicial presentan manchas oscuras pequeñas que son visualmente similares a imperfecciones naturales del fruto sano. Esto genera una alta tasa de falsos negativos (frutos enfermos clasificados como sanos), que es el error más costoso desde el punto de vista económico y sanitario.

**Solución aplicada:**
- Uso de **Dropout (30%)** en la arquitectura CNN para forzar al modelo a aprender características más generalizables y no depender de patrones superficiales.
- Selección del umbral de aceptación de F1-Score diferenciado para la clase A (≥ 0.60 vs ≥ 0.70 para las otras clases).

**Justificación:** El Dropout es una técnica de regularización que reduce el sobreajuste y mejora la capacidad del modelo para detectar características sutiles de enfermedades en etapas tempranas (Srivastava et al., 2014).

---

### Problema 3: Dataset relativamente pequeño para aprendizaje profundo

**Descripción:** Con 3,983 imágenes (resultado de aumentar 569 imágenes originales), el dataset es relativamente pequeño para entrenar una CNN desde cero con alta precisión. Los modelos de aprendizaje profundo generalmente requieren decenas de miles de imágenes para generalizar bien.

**Solución aplicada:**
- **Data Augmentation en tiempo de entrenamiento** mediante `ImageDataGenerator` de Keras: rotaciones ±20°, volteos horizontal y vertical, zoom 10%, desplazamientos.
- Consideración de **Transfer Learning** en el Entregable 3 (arquitecturas preentrenadas como VGG16, ResNet50) para aprovechar representaciones aprendidas en datasets grandes.

**Justificación:** El Transfer Learning ha demostrado alcanzar precisiones de 78.8% a 99.5% en clasificación de enfermedades en plantas con datasets pequeños (Qiao et al., 2017; Saleem et al., 2019).

---

### Problema 4: Ausencia de partición train/val/test predefinida

**Descripción:** El dataset original no incluye una separación predefinida entre conjuntos de entrenamiento, validación y prueba. Sin esta división, existe riesgo de data leakage (filtración de información del conjunto de prueba al entrenamiento) y de evaluaciones sesgadas del modelo.

**Solución aplicada:**
- División estratificada **80% entrenamiento / 10% validación / 10% prueba** con `train_test_split(stratify=...)` de scikit-learn.
- La estratificación garantiza que la proporción de cada clase sea igual en los tres subconjuntos, evitando particiones con distribuciones diferentes a las del dataset original.

**Justificación:** La división estratificada es una práctica estándar en Machine Learning para garantizar que el modelo se evalúe de forma justa y representativa en todas las clases.

---

### Problema 5: Imágenes originales de muy alta resolución

**Descripción:** Las imágenes originales del dataset fueron capturadas con resolución de 4,496 × 3,000 píxeles. Trabajar con imágenes de este tamaño requiere cantidades masivas de memoria RAM y tiempo de cómputo que exceden los recursos disponibles en Google Colab.

**Solución aplicada:**
- Redimensionamiento a **300 × 300 píxeles**, siguiendo el mismo criterio del artículo de referencia (Campos-Ferreira & González-Camacho, 2021).
- Las imágenes del dataset disponible ya están en 300 × 300 px, por lo que no es necesario realizar el redimensionamiento en el pipeline.

**Justificación:** El tamaño 300 × 300 px preserva suficiente detalle para identificar las características visuales de las enfermedades (textura, color, manchas) mientras reduce los requisitos computacionales en un factor de aproximadamente 150x respecto a la resolución original.

## 7. INGENIERIA DE CARACTERISTICAS

```python
# Importar librerias para ingeniería de caracteristicas
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight

print('Librerias de sklearn cargadas correctamente')
```

```python
# Label Encoding: convierte etiquetas de texto a valores numericos enteros
le = LabelEncoder()
df['label_encoded'] = le.fit_transform(df[columna_label])

print('Codificacion numerica de clases (Label Encoding):')
for clase, codigo in zip(le.classes_, range(len(le.classes_))):
    print(f'  {clase} -> {codigo}')

print()
df[[columna_label, 'label_encoded']].head(10)
```

```python
# Codificacion One-Hot para la capa de salida del modelo CNN
# Cada clase se representa como un vector binario
df_onehot = pd.get_dummies(df[columna_label], prefix='clase')
df_con_onehot = pd.concat([df, df_onehot], axis=1)

print('Codificacion One-Hot - vectores unitarios por clase:')
for clase in le.classes_:
    fila = df_con_onehot[df_con_onehot[columna_label] == clase].iloc[0]
    vec = [int(fila[f'clase_{c}']) for c in le.classes_]
    print(f'  {clase} -> {vec}')

print()
df_con_onehot[[columna_label, 'label_encoded'] + list(df_onehot.columns)].head(10)
```

```python
# Calcular pesos de clase para compensar el desbalance durante el entrenamiento
y = df['label_encoded'].values
pesos = compute_class_weight(class_weight='balanced', classes=np.unique(y), y=y)
pesos_dict = dict(zip(np.unique(y), pesos))

print('Pesos por clase para mitigar el desbalance:')
print('(Mayor peso = clase con menos muestras)')
print()
for codigo, clase in enumerate(le.classes_):
    n_muestras = df[df[columna_label] == clase].shape[0]
    peso = pesos_dict.get(codigo, 1.0)
    print(f'  Clase {clase}: {n_muestras} muestras  |  peso asignado = {peso:.4f}')
```

```python
# Division del dataset en conjuntos de entrenamiento, validacion y prueba
# 80% entrenamiento | 10% validacion | 10% prueba
# Se usa stratify para mantener la proporcion de clases en cada particion

if 'split' not in df.columns:
    df_train, df_temp = train_test_split(
        df, test_size=0.2, random_state=42, stratify=df[columna_label]
    )
    df_val, df_test = train_test_split(
        df_temp, test_size=0.5, random_state=42, stratify=df_temp[columna_label]
    )
else:
    df_train = df[df['split'] == 'train'].copy()
    df_val   = df[df['split'] == 'val'].copy()
    df_test  = df[df['split'] == 'test'].copy()

print('Division del dataset (con estratificacion):')
print(f'  Entrenamiento: {len(df_train)} registros ({len(df_train)/len(df)*100:.1f}%)')
print(f'  Validacion:    {len(df_val)} registros ({len(df_val)/len(df)*100:.1f}%)')
print(f'  Prueba:        {len(df_test)} registros ({len(df_test)/len(df)*100:.1f}%)')
```

```python
# Pipeline de preprocesamiento de imágenes para el modelo CNN
# NOTA: Las imágenes del dataset actual ya están en 300x300 px.
# El redimensionamiento se mantiene en el pipeline para garantizar compatibilidad
# con cualquier imagen nueva que se agregue al dataset en el futuro.
print('Pipeline de preprocesamiento de imágenes para el modelo CNN:')
print('=' * 65)
print()
print('Paso 1: Redimensionamiento')
print('   Tamaño objetivo:  300 x 300 px')
print('   Dataset actual:   Las imágenes ya están en 300x300 px')
print('   ℹ️  El redimensionamiento se mantiene en el pipeline como')
print('      salvaguarda para imágenes futuras que puedan tener')
print('      resoluciones diferentes (e.g., capturas nuevas en campo).')
print()
print('Paso 2: Normalización de valores de píxel')
print('   Rango original:    [0, 255]  (enteros)')
print('   Rango normalizado: [0.0, 1.0] (flotantes)')
print('   Fórmula:           pixel_normalizado = pixel / 255.0')
print('   Implementación:    rescale=1./255 en ImageDataGenerator')
print()
print('Paso 3: Data Augmentation (SOLO conjunto de entrenamiento)')
print('   Las siguientes transformaciones se aplican al vuelo durante')
print('   el entrenamiento. NO se generan archivos adicionales.')
print('   - Rotación aleatoria:        ±20 grados')
print('   - Volteo horizontal:         activado')
print('   - Volteo vertical:           activado')
print('   - Zoom aleatorio:            10%')
print('   - Desplazamiento horizontal: 10%')
print('   - Desplazamiento vertical:   10%')
print('   - Relleno de píxeles:        modo nearest')
print()
print('Paso 4: Sin augmentation para validación y prueba')
print('   Los conjuntos de validación y prueba SOLO reciben normalización.')
print('   Esto garantiza una evaluación limpia y objetiva del modelo.')
print()
print('Paso 5: Codificación de etiquetas (one-hot)')
print('   Healthy     → (1, 0, 0)')
print('   Scab        → (0, 1, 0)')
print('   Anthracnose → (0, 0, 1)')
print()
print('Paso 6: Organización en batches')
print('   Tamaño de batch: 128 imágenes')
print('   Número de épocas: 25')
print()
print('Código del pipeline (se implementará en Entregable 3):')
print()
print("""
from tensorflow.keras.preprocessing.image import ImageDataGenerator

IMG_SIZE   = (300, 300)   # Tamaño objetivo — el dataset actual ya está en 300x300,
                           # pero se especifica para estandarizar cualquier imagen futura
BATCH_SIZE = 128
EPOCHS     = 25

# Generador de entrenamiento CON augmentation
train_datagen = ImageDataGenerator(
    rescale=1./255,          # Normalización [0,255] → [0.0,1.0]
    rotation_range=20,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True,
    vertical_flip=True,
    zoom_range=0.1,
    fill_mode='nearest'
)

# Generador de validación y prueba SIN augmentation
val_test_datagen = ImageDataGenerator(rescale=1./255)

# target_size garantiza que todas las imágenes lleguen al modelo en 300x300,
# incluso si en el futuro se agregan imágenes con resolución diferente
train_gen = train_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=IMG_SIZE,    # Redimensiona si es necesario
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)
""")
```

La fase de preparación de datos permitió identificar cinco problemas concretos y definir soluciones técnicas justificadas para cada uno. A continuación se resume lo realizado y las decisiones tomadas:

**Problemas identificados y resueltos:**

1. **Desbalance de clases (Healthy: 44.2% / Scab: 30.0% / Anthracnose: 25.7%):** Se calcularon pesos de clase con `compute_class_weight='balanced'` de scikit-learn. Estos pesos se utilizarán en el parámetro `class_weight` del entrenamiento CNN en el Entregable 3, penalizando proporcionalmente más los errores en la clase Anthracnose, que es la menos representada.

2. **Similitud visual antracnosis temprana vs fruto sano:** Se estableció un umbral de aceptación diferenciado para la clase A (F1 ≥ 0.60) y se definirá Dropout (30%) en la arquitectura CNN para forzar al modelo a aprender características más discriminativas y no superficiales.

3. **Dataset relativamente pequeño:** Se configuró un pipeline de Data Augmentation con `ImageDataGenerator` que genera variaciones (rotaciones, volteos, zoom) al vuelo durante el entrenamiento, sin necesidad de imágenes adicionales. En el Entregable 3 se evaluará Transfer Learning como estrategia complementaria.

4. **Ausencia de partición predefinida:** Se realizó una división estratificada **80% / 10% / 10%** (entrenamiento / validación / prueba) con `train_test_split(stratify=Condition)`, garantizando que las proporciones de cada clase sean representativas en los tres subconjuntos. Los archivos CSV resultantes se guardaron en Google Drive para su uso en el Entregable 3.

5. **Alta resolución de imágenes originales:** Las imágenes del dataset disponible ya están en 300 × 300 px (resolución procesada del estudio original), por lo que no se requiere redimensionamiento adicional. El pipeline de preprocesamiento solo aplica normalización al rango [0, 1].

**Decisiones técnicas clave:**
- Codificación de etiquetas en dos formatos: Label Encoding (para cálculo de pesos) y One-Hot Encoding (para la capa de salida del CNN).
- División 80/10/10 elegida sobre 70/15/15 para maximizar los datos de entrenamiento dado el tamaño moderado del dataset.
- Se reservó el conjunto de prueba exclusivamente para la evaluación final del modelo, sin utilizarlo en ninguna decisión de diseño o ajuste de hiperparámetros.

**Conclusión personal:** Esta fase reveló que la mayor complejidad del proyecto no reside en la calidad de los datos (que es excelente) sino en las características inherentes al problema: el desbalance de clases y la dificultad visual de distinguir antracnosis temprana de frutos sanos. Las estrategias definidas aquí —pesos de clase, Data Augmentation, división estratificada— son la base para que el modelo CNN del Entregable 3 pueda superar el umbral de 80% de precisión global y alcanzar un F1-Score aceptable para las tres clases objetivo.
